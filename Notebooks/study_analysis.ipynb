# Study Log Analysis

This notebook analyzes my self-recorded study time during exam preparation.

The goal is to understand how my study time is distributed across subjects
and identify potential imbalance or focus patterns.

```python
import pandas as pd

df = pd.read_csv("../Data/study_log.csv")
df
```

## Load dataset

```python
df["date"] = pd.to_datetime(df["date"])
df.dtypes
```

```python
subject_hours = (
    df.groupby("subject")["hours"]
    .sum()
    .sort_values(ascending=False)
)

subject_hours
```

```python
## Aggregate study hours by subject
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
subject_hours.plot(kind="bar")
plt.title("Total Study Hours by Subject")
plt.ylabel("Hours")
plt.xlabel("Subject")
plt.tight_layout()

plt.savefig("../figures/subject_breakdown.png")
plt.show()
```

## Visualization: total study hours by subject

```python
daily_hours = (
    df.groupby("date")["hours"]
      .sum()
)

plt.figure(figsize=(6,4))
daily_hours.plot(marker="o")
plt.title("Daily Study Time Trend")
plt.ylabel("Hours")
plt.xlabel("Date")
plt.tight_layout()

plt.savefig("../figures/daily_trend.png")
plt.show()
```

## Visualization: daily study hours by subject

```python
# Pivot data to get daily study hours per subject
daily_subject_hours = df.pivot_table(
    index="date",
    columns="subject",
    values="hours",
    aggfunc="sum",
    fill_value=0
)

daily_subject_hours
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 4))
daily_subject_hours.plot(kind="bar", stacked=True)

plt.title("Daily Study Hours by Subject")
plt.xlabel("Date")
plt.ylabel("Hours")
plt.tight_layout()

plt.savefig("../figures/daily_subject_breakdown.png")
plt.show()
```

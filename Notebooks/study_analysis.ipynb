# Study Log Analysis

This notebook analyzes my self-recorded study time during exam preparation.

The goal is to understand how my study time is distributed across subjects
and identify potential imbalance or focus patterns.

```python
import pandas as pd

df = pd.read_csv("../Data/study_log.csv")
df
```

## Load dataset

```python
df["date"] = pd.to_datetime(df["date"])
df.dtypes
```

```python
subject_hours = (
    df.groupby("subject")["hours"]
    .sum()
    .sort_values(ascending=False)
)

subject_hours
```

```python
## Aggregate study hours by subject
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
subject_hours.plot(kind="bar")
plt.title("Total Study Hours by Subject")
plt.ylabel("Hours")
plt.xlabel("Subject")
plt.tight_layout()

plt.savefig("../figures/subject_breakdown.png")
plt.show()
```

## Visualization: total study hours by subject

```python
daily_hours = (
    df.groupby("date")["hours"]
      .sum()
)

plt.figure(figsize=(6,4))
daily_hours.plot(marker="o")
plt.title("Daily Study Time Trend")
plt.ylabel("Hours")
plt.xlabel("Date")
plt.tight_layout()

plt.savefig("../figures/daily_trend.png")
plt.show()
```

## Analysis Summary

From the analysis of my self-recorded study logs, several patterns can be observed:

1. **Subject-level distribution**
   - Discrete Mathematics accounts for the largest share of total study time.
   - English and Advanced Programming receive relatively balanced but lower attention.
   - This reflects my current priority on strengthening mathematical foundations.

2. **Daily study trend**
   - Daily total study time fluctuates between 3–4 hours.
   - A noticeable increase appears toward the end of the observed period, indicating improved consistency.

3. **Limitations**
   - The dataset is small and covers only a short time window.
   - Future analysis could benefit from longer-term data and additional variables such as difficulty or fatigue.

Overall, this analysis demonstrates basic data cleaning, aggregation, and visualization skills, as well as the ability to interpret results in context.
